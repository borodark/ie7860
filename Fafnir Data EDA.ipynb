```python
import pandas as pd
import numpy as np
import itertools
```

```python
data = pd.HDFStore('bdv-data.h5')
print(data.keys())
```

```python
gifi = data['gifi']
gifi.sort_index(inplace=True)
gby = gifi.groupby(level='year')
```

```python
def filter_by_list_of_yrs(gifi_df, years=[2011,2012,2013,2014,2015,2016,2017,2018,2019]):
    gby = gifi_df.groupby(level='year')
    bvd_ids = None
    for y in years:
        current_year_bv_ids = set(gby.groups[y].droplevel('year'))
        if bvd_ids == None:
            bvd_ids = current_year_bv_ids
        else:
            bvd_ids = bvd_ids.intersection(current_year_bv_ids)
    tuples = list(itertools.product(bvd_ids,years))
    select_index = pd.MultiIndex.from_tuples(tuples, names=['bv_id','year'])
    rc = gifi_df.loc[select_index]
    rc.sort_index(inplace=True)
    return rc
```

```python
all9 = filter_by_list_of_yrs(gifi)
all9.head(36)
```
