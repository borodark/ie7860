```python
import pandas as pd
```

```python
gifi = pd.HDFStore('bdv-gifi-all.h5')
size_tuples = []
for gifi_column_key in gifi.keys():
    column_df = gifi.get(gifi_column_key)
    size_tuples.append((gifi_column_key, column_df.size))
    

top_columns = sorted(size_tuples, key=lambda col_values: -1 * col_values[1])
top_columns
```

```python
column_df = None
joined_df = None
i=0
for (gifi_column_key, size) in top_columns:
    column_df = gifi.get(gifi_column_key)
    print(column_df.memory_usage(index=True))
    if joined_df is None:
        joined_df = column_df
    else:
        joined_df = joined_df.join(column_df,on=['bv_id','year'], how='outer')
        joined_df.info(memory_usage=True)
        #joined_df.memory_usage(index=True)
    i=i+1
    print(str(i))
    if i > 3:
        break
gifi_joined = pd.HDFStore('bdv-gifi-3.h5')
gifi_joined.put('gifi', joined_df)
gifi_joined.close()
```

```python
column_df = None
net_assets_th_gbp_turnover_th_gbp = None
i=0
for gifi_column_key in ['net_assets_th_gbp','turnover_th_gbp']:
    column_df = gifi.get(gifi_column_key)
    print(column_df.memory_usage(index=True))
    if net_assets_th_gbp_turnover_th_gbp is None:
        net_assets_th_gbp_turnover_th_gbp = column_df
    else:
        net_assets_th_gbp_turnover_th_gbp = net_assets_th_gbp_turnover_th_gbp.join(column_df,on=['bv_id','year'], how='inner')
        net_assets_th_gbp_turnover_th_gbp.info(memory_usage=True)
        #joined_df.memory_usage(index=True)
    i=i+1
    print(str(i))
    if i > 3:
        break
```
