```python
import pandas as pd
import numpy as np
import re
```

```python
gifi = pd.HDFStore('bdv-gifi-all.h5')
size_tuples = []
for gifi_column_key in gifi.keys():
    column_df = gifi.get(gifi_column_key)
    size_tuples.append((gifi_column_key, column_df.size))
    

top_columns = sorted(size_tuples, key=lambda col_values: -1 * col_values[1])
top_columns
```

```python
talcl  = gifi.get('total_assets_less_cur._liab._th_gbp')
talcl.reset_index(inplace=True)
print(talcl.shape)
talcl.drop_duplicates(subset=['bv_id','year'], inplace=True)
print(talcl.shape)
df1 =talcl
```

```python
net_assets_th_gbp  = gifi.get('net_assets_th_gbp')
net_assets_th_gbp.reset_index(inplace=True)# to_csv('net_assets_th_gbp.csv')
print(net_assets_th_gbp.shape)
net_assets_th_gbp.drop_duplicates(subset=['bv_id','year'], inplace=True)
print(net_assets_th_gbp.shape)
#net_assets_th_gbp
```

```python
data = pd.HDFStore('bdv-data.h5')
print(data.keys())
```

```python
column_df = None
joined_df = None
i=0
for (gifi_column_key, size) in top_columns:
    column_df = gifi.get(gifi_column_key).reset_index()
    column_df.drop_duplicates(subset=['bv_id','year'], inplace=True,ignore_index=True)
    print("Column shape going into merge " + str(column_df.shape))
    if joined_df is None:
        joined_df = column_df
    else:
        print("Joined shape going onto merge " + str(joined_df.shape))
        joined_df = pd.merge(joined_df,column_df,on=['bv_id','year'], how='outer')
        print("Joined shape after merge " + str(joined_df.shape))
        joined_df.drop_duplicates(subset=['bv_id','year'], inplace=True,ignore_index=True)
    joined_df.info(memory_usage=True)
    i=i+1
    print(str(i))

joined_df.set_index(['bv_id','year'],inplace=True)
data['gifi'] = joined_df
data.close()
joined_df.to_csv('gifi.csv')
```
