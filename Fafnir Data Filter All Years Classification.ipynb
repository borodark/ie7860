```python
import pandas as pd
import numpy as np
```

```python
data = pd.HDFStore('bdv-data.h5')
print(data.keys())
```

```python
gifi9 = data['all_nine_years']
gifi9.head(32)
data.close()
```

```python
gbi = gifi9.groupby(level='bv_id')
gbi.groups
```

```python
yZ = ['solvency_ratio_(asset_based)_(%)',
'solvency_ratio_(liability_based)_(%)',
'liquidity_ratio_(x)',
'shareholders_liquidity_ratio_(x)']
```

```python

for y in yZ:
    gifi9["UP_"+ y] = pd.Series(False, index=gifi9.index)
    years = gifi9[y]
    shifted = pd.DataFrame(years).shift(periods=-1)
    gifi9["UP_"+ y] = shifted.subtract(years, axis=0) > 0
```

```python
gifi9.drop(index=2019, level=1, inplace=True)
gifi9.columns
```
