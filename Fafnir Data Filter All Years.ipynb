```python
import pandas as pd
import numpy as np
```

```python
data = pd.HDFStore('bdv-data.h5')
print(data.keys())
```

```python

print(data['flat'].shape)
print(data['gifi'].shape)
```

```python
gifi = data['gifi']
gifi.sort_index(inplace=True)
gifi.head(32)
```

```python
gby = gifi.groupby(level='year')
gby.groups
```

```python
_2011 = gby.groups[2011].droplevel('year')
print(_2011.shape)
_2015 = gby.groups[2015].droplevel('year')
print(_2015.shape)
_i = set(_2011).intersection(set(_2015))
len(_i)
```

```python
y_all = [2011,2012,2013,2014,2015,2016,2017,2018,2019]
all_9 = None
for y in y_all:
    current_year_bv_ids = set(gby.groups[y].droplevel('year'))
    if all_9 == None:
        all_9 = current_year_bv_ids
    else:
        all_9 = all_9.intersection(current_year_bv_ids)
len(all_9)
```

```python
import itertools
bvd_ids = all_9
tuples = list(itertools.product(bvd_ids,y_all))
select_index = pd.MultiIndex.from_tuples(tuples, names=['bv_id','year'])
select_index
```

```python
all_nine_y = gifi.loc[select_index]
all_nine_y.sort_index(inplace=True)
all_nine_y.describe()
```

```python
all_nine_y.head(36)
```

```python
#all_nine_y.to_csv('all_nine_years.csv')
data['all_nine_years'] = all_nine_y
data.close()
```
